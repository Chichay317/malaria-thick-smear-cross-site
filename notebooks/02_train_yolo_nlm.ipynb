```python
!pip install -q ultralytics
import ultralytics, yaml
from pathlib import Path
print("ultralytics", ultralytics.__version__)

src = next(Path("/kaggle/input").rglob("nlm_tiles"))        
print("Tiles found at:", src)
for sp in ["train", "val", "test"]:
    print(sp, len(list((src / sp / "images").glob("*.jpg"))), "tiles")

yaml.safe_dump({"path": str(src), "train": "train/images", "val": "val/images", "test": "test/images",
                "names": {0: "parasite", 1: "wbc"}}, open("/kaggle/working/data.yaml", "w"))
```

```python
from ultralytics import YOLO
model = YOLO("yolo11s.pt")
model.train(data="/kaggle/working/data.yaml", epochs=40, patience=15, imgsz=640, batch=32,
            device=0, workers=4, seed=0, project="/kaggle/working/runs", name="nlm_y11s",
            exist_ok=True, plots=True)
```

```python
import pandas as pd
best = YOLO("/kaggle/working/runs/nlm_y11s/weights/best.pt")
m = best.val(data="/kaggle/working/data.yaml", split="val", imgsz=640, batch=32, device=0, plots=False)
for i, name in enumerate(["parasite", "wbc"]):
    print(f"{name:9s}  precision {m.box.p[i]:.3f}  recall {m.box.r[i]:.3f}  mAP50 {m.box.ap50[i]:.3f}")

hist = pd.read_csv("/kaggle/working/runs/nlm_y11s/results.csv")
hist.columns = hist.columns.str.strip()
print("\nRounds trained:", len(hist))
print(hist[["epoch", "metrics/precision(B)", "metrics/recall(B)", "metrics/mAP50(B)"]].tail(5).round(3).to_string(index=False))
```
