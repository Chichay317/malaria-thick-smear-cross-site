```python
!pip install -q ultralytics
import yaml
from pathlib import Path
from ultralytics import YOLO
inp = Path("/kaggle/input")
S = {"bd": next(p for p in inp.rglob("nlm_tiles") if p.is_dir()),
     "gh": next(p for p in inp.rglob("gh_split") if p.is_dir()),
     "tz": next(p for p in inp.rglob("tz_split") if p.is_dir())}
for k, v in S.items():
    print(k, v, "| train tiles:", len(list((v / "train" / "images").glob("*.jpg"))))

def run(name, sites):
    y = f"/kaggle/working/{name}.yaml"
    yaml.safe_dump({"train": [str(S[s] / "train" / "images") for s in sites],
                    "val": [str(S[s] / "val" / "images") for s in sites],
                    "names": {0: "parasite", 1: "wbc"}}, open(y, "w"))
    try:  
        YOLO("yolo11s.pt").train(data=y, epochs=40, patience=15, imgsz=640, batch=32, device=0, workers=4,
                                 seed=0, project="/kaggle/working/runs", name=name, exist_ok=True, plots=False)
    except Exception as e:
        print(f"RUN {name} FAILED:", e)
```

```python
   run("B_gh", ["gh"])
   run("C_tz", ["tz"])
   run("D_gh_tz", ["gh", "tz"])
```
